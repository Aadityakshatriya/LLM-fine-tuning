```python
import torch
import torch.nn.functional as F
import io
import soundfile as sf
import os
from typing import Any, Dict, List, Union
from dataclasses import dataclass

from datasets import load_dataset, Features, Value
from torch.utils.data import DataLoader as TorchLoader # Your custom loader
from transformers import (
    WhisperProcessor, 
    WhisperForConditionalGeneration,
    Seq2SeqTrainingArguments, 
    Seq2SeqTrainer
)
import evaluate
```

```python
# --- 1. CONFIGURATION ---
MODEL_ID = "openai/whisper-small" 
LANGUAGE_ABBREVIATION = "hi" # Hindi
TASK = "transcribe"
OUTPUT_DIR = "./whisper-small-hindi-finetuned"
BATCH_SIZE = 4 # Start conservative on batch size for CPU training
TARGET_SAMPLING_RATE = 16000 # Whisper's required rate
```

```python
# --- 2. YOUR CUSTOM DATA WRAPPER ---
# Using the HFDataWrapper class from your notebook
class HFDataWrapper:
    def __init__(self, ds_split):
        self.ds = ds_split

    def __len__(self):
        return len(self.ds)

    def __getitem__(self, idx):
        sample = self.ds[idx]
        audio_info = sample.get("audio", sample.get("audio"))

        # Robust audio decoding: support 'bytes', 'array', or 'path'
        waveform = None
        sr = None
        try:
            if isinstance(audio_info, dict):
                # 1) bytes field (raw binary)
                if audio_info.get("bytes") is not None:
                    raw = audio_info.get("bytes")
                    raw = bytes(raw)  # ensure bytes (handles memoryview)
                    with io.BytesIO(raw) as f:
                        waveform, sr = sf.read(f, dtype="float32")
                # 2) array field (datasets Audio() after decoding)
                elif audio_info.get("array") is not None:
                    arr = audio_info.get("array")
                    waveform = arr
                    sr = audio_info.get("sampling_rate", TARGET_SAMPLING_RATE)
                # 3) path string pointing to a file
                elif audio_info.get("path"):
                    path = audio_info.get("path")
                    waveform, sr = sf.read(path, dtype="float32")
            else:
                # If audio_info is already an array-like
                if hasattr(audio_info, "shape") or isinstance(audio_info, (list, tuple)):
                    waveform = audio_info
                    sr = TARGET_SAMPLING_RATE
        except Exception as e:
            raise RuntimeError(f"Failed to decode audio for idx={idx}: {e}") from e

        if waveform is None:
            raise ValueError(f"No valid audio found in sample index {idx}: {audio_info}")

        # convert to torch tensor and ensure shape (channels, time)
        wav = waveform
        if isinstance(wav, list):
            import numpy as _np
            wav = _np.array(wav, dtype="float32")
        if hasattr(wav, "ndim") and wav.ndim == 1:
            wav = torch.tensor(wav).unsqueeze(0)
        else:
            # If shape is (time, channels) -> transpose to (channels, time)
            wav = torch.tensor(wav.T) if hasattr(wav, "ndim") and wav.ndim == 2 else torch.tensor(wav)

        meta = {k: v for k, v in sample.items() if k != "audio"}

        # Robust transcript extraction: try common keys and coerce to str
        transcript = None
        for key in ("transcript", "transcription", "text", "sentence", "label"):
            if key in meta and meta[key] is not None:
                transcript = meta[key]
                break
        if transcript is None:
            # fallback: try 'utterance' or any other field that looks like text
            for k, v in meta.items():
                if isinstance(v, str) and len(v) > 0:
                    transcript = v
                    break
        if transcript is None:
            transcript = ""
        # If transcript is a list, join into a single string
        if isinstance(transcript, (list, tuple)):
            transcript = " ".join(map(str, transcript))
        # Ensure string type
        transcript = str(transcript)

        return wav, sr, transcript
```

```python
from dataclasses import dataclass
import torch
import torch.nn.functional as F
from typing import Any, Dict, List

TARGET_SAMPLING_RATE = 16000  # same as in your code
TARGET_MEL_FRAMES = 3000      # same as in ASRPipeline


@dataclass
class CustomTrainingDataCollator:
    processor: WhisperProcessor

    def __call__(self, batch: List[Any]) -> Dict[str, torch.Tensor]:
        # batch: [(waveform, sr, text), ...] from HFDataWrapper
        waveforms, srs, texts = zip(*batch)

        # ---- 1. Make waveforms like your collate_batch ----
        # Ensure shape (time,)
        waveforms = [w.squeeze() for w in waveforms]

        # Pad raw audio to max length in this batch
        max_len = max(w.shape[-1] for w in waveforms)
        padded = [F.pad(w, (0, max_len - w.shape[-1])) for w in waveforms]
        batch_waveforms = torch.stack(padded)  # (batch, time)

        # ASRPipeline expects (batch, 1, time) sometimes, but your feature_extractor
        # works fine with list of 1D arrays. Let's do that:
        wf_list = [w.cpu().numpy() for w in batch_waveforms]  # list[np.ndarray]

        # Use first sample rate (all are 16000 anyway)
        sr = int(srs[0])

        # ---- 2. Extract mel spectrograms exactly like ASRPipeline ----
        inputs = self.processor.feature_extractor(
            wf_list,
            sampling_rate=sr,
            return_tensors="pt",
            padding="longest"
        )
        input_features = inputs.input_features  # (B, 80, frames)

        B, n_mels, frames = input_features.shape
        target_len = TARGET_MEL_FRAMES

        # Manually pad / truncate just like in ASRPipeline
        if frames < target_len:
            pad_amt = target_len - frames
            input_features = F.pad(input_features, (0, pad_amt))
        elif frames > target_len:
            input_features = input_features[..., :target_len]

        # Safety: ASSERT that we really have 3000 frames now
        assert input_features.shape[-1] == target_len, f"Got mel length {input_features.shape[-1]}"

        # ---- 3. Tokenize labels (same idea as before) ----
        str_labels = [s if isinstance(s, str) else str(s) for s in texts]
        tokenized = self.processor.tokenizer(
            str_labels,
            max_length=448,
            padding="longest",
            truncation=True,
            return_tensors="pt",
        )

        labels = tokenized["input_ids"].masked_fill(
            tokenized.attention_mask.ne(1), -100
        )

        return {
            "input_features": input_features,
            "labels": labels,
        }
```

```python
# --- 4. DATA LOADING ---

print("Step 1: Loading Dataset...")

# Load the ARTPARK-IISc/vaani dataset
# NOTE: Ensure you are authenticated/logged in to HF for this public dataset.
try:
    ds = load_dataset("ARTPARK-IISc/vaani", "Goa_NorthSouthGoa", split="train")
except:
    print("🚨 Dataset load failed. Ensure you have the dataset cached locally or are authenticated.")
    # Fallback to local file paths if necessary, but this assumes your notebook was running on the HF dataset.
    raise

# Disable auto-decoding to get raw bytes
ds = ds.cast_column("audio", Features({"path": Value("string"), "bytes": Value("binary")}))

ds = ds.filter(lambda x: x["language"] == "Hindi")

# Split the Hugging Face Dataset before wrapping
ds_splits = ds.train_test_split(test_size=0.05)
train_ds = ds_splits["train"]
eval_ds = ds_splits["test"]

# Wrap the datasets with your custom decoder
train_wrapped_ds = HFDataWrapper(train_ds)
eval_wrapped_ds = HFDataWrapper(eval_ds)

# Quick diagnostics to catch bad transcripts/audio early
print("Wrapped dataset sizes:", len(train_wrapped_ds), len(eval_wrapped_ds))
for i in range(min(3, len(train_wrapped_ds))):
    try:
        wf, sr, txt = train_wrapped_ds[i]
        print(f"Sample {i}: sr={sr}, waveform shape={getattr(wf, 'shape', None)}, transcript type={type(txt)}, snippet={repr(txt)[:120]}")
    except Exception as e:
        print(f"Sample {i} diagnostic failed: {e}")
```

```python
# --- 5. PROCESSOR, MODEL, AND TRAINER SETUP ---

print("Step 2: Initializing Processor, Model, and Trainer...")

processor = WhisperProcessor.from_pretrained(
    MODEL_ID, 
    language=LANGUAGE_ABBREVIATION, 
    task=TASK
)

# Initialize the data collator
data_collator = CustomTrainingDataCollator(processor=processor)

# Load the model
model = WhisperForConditionalGeneration.from_pretrained(MODEL_ID)
model.config.forced_decoder_ids = processor.get_decoder_prompt_ids(
    language=LANGUAGE_ABBREVIATION, 
    task=TASK
)
model.config.suppress_tokens = []
model.config.use_cache = False

# Load evaluation metric
metric = evaluate.load("wer")

# --- Real-time monitoring state and callback (defines RealtimeMonitoringCallback) ---
import matplotlib.pyplot as plt
import matplotlib
matplotlib.use('Agg')
import numpy as np
from collections import defaultdict
import json
import time
import os

MONITOR_DIR = "./training_monitor"
os.makedirs(MONITOR_DIR, exist_ok=True)

training_metrics = {
    "steps": [],
    "train_loss": [],
    "eval_loss": [],
    "eval_wer": [],
    "timestamps": []
}

data_analysis = {
    "duration_buckets": defaultdict(lambda: {"count": 0, "wer_samples": []}),
    "length_buckets": defaultdict(lambda: {"count": 0, "wer_samples": []}),
    "snr_values": [],
    "repetition_scores": [],
    "speech_ratios": [],
}

from transformers import TrainerCallback


class RealtimeMonitoringCallback(TrainerCallback):
    def __init__(self, eval_dataset, processor, model, config):
        self.eval_dataset = eval_dataset
        self.processor = processor
        self.model = model
        self.config = config
        self.step_count = 0

    def on_log(self, args, state, control, logs=None, **kwargs):
        if logs is None:
            return
        from datetime import datetime
        self.step_count = state.global_step

        if "loss" in logs:
            training_metrics["train_loss"].append(logs["loss"])
        if "eval_loss" in logs:
            training_metrics["eval_loss"].append(logs["eval_loss"])
        if "eval_wer" in logs:
            training_metrics["eval_wer"].append(logs["eval_wer"])

        training_metrics["steps"].append(self.step_count)
        training_metrics["timestamps"].append(datetime.now().isoformat())

        self._generate_plots()
        self._print_summary(logs)
        self._save_metrics()

    def _generate_plots(self):
        if len(training_metrics["steps"]) == 0:
            return
        fig, axes = plt.subplots(2, 2, figsize=(14, 10))
        fig.suptitle(f'Real-Time Training Monitor (Step {self.step_count})', fontsize=16, fontweight='bold')

        ax = axes[0, 0]
        if training_metrics["train_loss"]:
            ax.plot(training_metrics["steps"][:len(training_metrics["train_loss"])], training_metrics["train_loss"], label='Train Loss', marker='o', linewidth=2)
        if training_metrics["eval_loss"]:
            ax.plot(training_metrics["steps"][:len(training_metrics["eval_loss"])], training_metrics["eval_loss"], label='Eval Loss', marker='s', linewidth=2, color='red')
        ax.set_xlabel('Step')
        ax.set_ylabel('Loss')
        ax.set_title('📊 Loss Trajectory')
        ax.legend()
        ax.grid(True, alpha=0.3)

        ax = axes[0, 1]
        if training_metrics["eval_wer"]:
            ax.plot(training_metrics["steps"][:len(training_metrics["eval_wer"])], training_metrics["eval_wer"], label='WER', marker='D', linewidth=2, color='green')
            ax.fill_between(training_metrics["steps"][:len(training_metrics["eval_wer"])], training_metrics["eval_wer"], alpha=0.3, color='green')
        ax.set_xlabel('Step')
        ax.set_ylabel('WER (%)')
        ax.set_title('📈 Word Error Rate')
        ax.legend()
        ax.grid(True, alpha=0.3)

        ax = axes[1, 0]
        if data_analysis["snr_values"]:
            ax.hist(data_analysis["snr_values"], bins=20, alpha=0.7, label='SNR (dB)', color='blue')
            ax.axvline(np.mean(data_analysis["snr_values"]), color='blue', linestyle='--', linewidth=2, label=f'Avg SNR: {np.mean(data_analysis["snr_values"]):.1f}')
        if data_analysis["speech_ratios"]:
            ax2 = ax.twinx()
            ax2.hist([r*100 for r in data_analysis["speech_ratios"]], bins=20, alpha=0.5, label='Speech Ratio (%)', color='orange')
            ax2.set_ylabel('Frequency (Speech Ratio)')
            ax2.legend(loc='upper right')
        ax.set_xlabel('SNR / Speech Ratio')
        ax.set_title('🎤 Audio Quality Metrics')
        ax.legend(loc='upper left')

        ax = axes[1, 1]
        if data_analysis["repetition_scores"]:
            ax.hist(data_analysis["repetition_scores"], bins=20, alpha=0.7, label='Repetition Score', color='red')
            ax.axvline(np.mean(data_analysis["repetition_scores"]), color='red', linestyle='--', linewidth=2, label=f'Avg: {np.mean(data_analysis["repetition_scores"]):.2f}')
        ax.set_xlabel('Repetition Score')
        ax.set_ylabel('Frequency')
        ax.set_title('⚠️ Hallucination/Repetition Detection')
        ax.legend()
        ax.grid(True, alpha=0.3)

        plt.tight_layout()
        plt.savefig(f"{MONITOR_DIR}/training_plots_step_{self.step_count}.png", dpi=100, bbox_inches='tight')
        plt.savefig(f"{MONITOR_DIR}/latest_plots.png", dpi=100, bbox_inches='tight')
        plt.close()

    def _print_summary(self, logs):
        print("\n" + "=" * 90)
        print(f"⏱️  STEP {self.step_count} | REAL-TIME TRAINING MONITOR")
        print("=" * 90)
        if "loss" in logs:
            print(f"  • Train Loss:        {logs['loss']:.4f}")
        if "eval_loss" in logs:
            print(f"  • Validation Loss:   {logs['eval_loss']:.4f}")
            if len(training_metrics["eval_loss"]) > 1:
                delta = training_metrics["eval_loss"][-1] - training_metrics["eval_loss"][-2]
                trend = "📉 Improving" if delta < 0 else "📈 Degrading"
                print(f"    └─ Change: {delta:+.4f} ({trend})")
        print("\n📈 WER METRICS:")
        if "eval_wer" in logs:
            print(f"  • Current WER:       {logs['eval_wer']:.2f}%")
            if len(training_metrics["eval_wer"]) > 1:
                delta = training_metrics["eval_wer"][-1] - training_metrics["eval_wer"][-2]
                trend = "✅ Improving" if delta < 0 else "⚠️  Degrading"
                print(f"    └─ Change: {delta:+.2f}% ({trend})")
            if len(training_metrics["eval_wer"]) > 0:
                best_wer = min(training_metrics["eval_wer"])
                print(f"  • Best WER (so far):  {best_wer:.2f}%")
        print("\n🎤 DATA CHARACTERISTICS:")
        if data_analysis["snr_values"]:
            print(f"  • Avg SNR:           {np.mean(data_analysis['snr_values']):.2f} dB (Range: {np.min(data_analysis['snr_values']):.1f} - {np.max(data_analysis['snr_values']):.1f})")
        if data_analysis["speech_ratios"]:
            print(f"  • Avg Speech Ratio:  {np.mean(data_analysis['speech_ratios']):.1%}")
        if data_analysis["repetition_scores"]:
            avg_rep = np.mean(data_analysis["repetition_scores"])
            status = "⚠️  HIGH" if avg_rep > 0.15 else "✅ OK"
            print(f"  • Avg Repetition:    {avg_rep:.3f} ({status})")
        print("\n📦 DURATION BUCKETS (samples analyzed):")
        for bucket, data in sorted(data_analysis["duration_buckets"].items()):
            print(f"  • {bucket:8}: {data['count']:3} samples")
        print("\n📝 TRANSCRIPT LENGTH BUCKETS:")
        for bucket, data in sorted(data_analysis["length_buckets"].items()):
            print(f"  • {bucket:8}: {data['count']:3} samples")
        print("\n" + "=" * 90)

    def _save_metrics(self):
        metrics_file = f"{MONITOR_DIR}/metrics.json"
        with open(metrics_file, 'w') as f:
            json.dump(training_metrics, f, indent=2)


# Enhanced compute_metrics with data analysis
def compute_metrics(pred):
    import numpy as np
    from collections import defaultdict
    
    pred_ids = pred.predictions
    label_ids = pred.label_ids

    label_ids[label_ids == -100] = processor.tokenizer.pad_token_id

    pred_str = processor.tokenizer.batch_decode(pred_ids, skip_special_tokens=True)
    label_str = processor.tokenizer.batch_decode(label_ids, skip_special_tokens=True)

    wer = 100 * metric.compute(predictions=pred_str, references=label_str)
    
    # Collect data characteristics for monitoring
    try:
        for i, sample_idx in enumerate(pred.label_ids.shape[0:1]):
            wav, sr, txt = eval_wrapped_ds[i]
            duration = wav.shape[-1] / sr
            text_len = len(txt.split())
            
            # Duration buckets
            if duration < 2:
                bucket = "<2s"
            elif duration < 5:
                bucket = "2-5s"
            elif duration < 10:
                bucket = "5-10s"
            elif duration < 20:
                bucket = "10-20s"
            else:
                bucket = ">20s"
            data_analysis["duration_buckets"][bucket]["count"] += 1
            
            # Length buckets
            if text_len <= 10:
                bucket = "1-10"
            elif text_len <= 25:
                bucket = "11-25"
            elif text_len <= 50:
                bucket = "26-50"
            elif text_len <= 100:
                bucket = "51-100"
            else:
                bucket = ">100"
            data_analysis["length_buckets"][bucket]["count"] += 1
            
            # SNR estimation
            wav_np = wav.cpu().numpy().flatten()
            threshold = np.std(wav_np) * 0.1
            silence_power = np.mean(wav_np[np.abs(wav_np) < threshold] ** 2)
            signal_power = np.mean(wav_np ** 2)
            snr = 10 * np.log10(max(signal_power / max(silence_power, 1e-10), 1))
            data_analysis["snr_values"].append(snr)
            
            # Speech ratio
            speech_frames = np.sum(np.abs(wav_np) > threshold)
            speech_ratio = speech_frames / len(wav_np)
            data_analysis["speech_ratios"].append(speech_ratio)
            
            # Repetition detection
            words = txt.lower().split()
            repetitions = 0
            window = 3
            for j in range(len(words) - window):
                seq = ' '.join(words[j:j+window])
                count = sum(1 for k in range(j+window, len(words)-window+1) 
                           if ' '.join(words[k:k+window]) == seq)
                if count > 0:
                    repetitions += count
            rep_score = repetitions / max(len(words) - window, 1)
            data_analysis["repetition_scores"].append(rep_score)
    except:
        pass

    return {"wer": wer}


# Define training arguments (train by epochs instead of fixed max_steps)
training_args = Seq2SeqTrainingArguments(
    output_dir=OUTPUT_DIR,
    per_device_train_batch_size=BATCH_SIZE,
    gradient_accumulation_steps=1, 
    learning_rate=1e-5,
    warmup_steps=500,
    num_train_epochs=3,  # <-- run 3 full epochs
    eval_strategy="steps",
    per_device_eval_batch_size=BATCH_SIZE,
    predict_with_generate=True,
    generation_max_length=225,
    save_steps=1000,
    eval_steps=1000,
    logging_steps=25,
    report_to=["tensorboard"],
    load_best_model_at_end=True,
    metric_for_best_model="wer",
    greater_is_better=False, 
)

# Print estimated steps per epoch and total steps for visibility
steps_per_epoch = (len(train_ds) + training_args.per_device_train_batch_size - 1) // training_args.per_device_train_batch_size
total_estimated_steps = steps_per_epoch * training_args.num_train_epochs
print(f"Training for {training_args.num_train_epochs} epochs -> approx {steps_per_epoch} steps/epoch, {total_estimated_steps} total steps")

# Initialize the Trainer
trainer = Seq2SeqTrainer(
    model=model,
    args=training_args,
    # Pass your custom wrapped datasets here:
    train_dataset=train_wrapped_ds,
    eval_dataset=eval_wrapped_ds,
    data_collator=data_collator,
    compute_metrics=compute_metrics,
    tokenizer=processor,
)

# Add real-time monitoring callback
realtime_callback = RealtimeMonitoringCallback(
    eval_dataset=eval_wrapped_ds,
    processor=processor,
    model=model,
    config=training_args
)
trainer.add_callback(realtime_callback)

# --- 6. START TRAINING ---
print("-" * 60)
print("Step 3: Starting Fine-Tuning...")
print(f"📊 Live plots will be saved to: {MONITOR_DIR}")
print(f"🔄 Check {MONITOR_DIR}/latest_plots.png for real-time updates")
print("-" * 60)

trainer.train()

print("-" * 60)
print(f"✅ Training Complete! Model checkpoints saved to {OUTPUT_DIR}")
print(f"📊 Training plots saved to {MONITOR_DIR}")
print("-" * 60)
```
