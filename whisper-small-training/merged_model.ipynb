```python
import torch
from transformers import WhisperForConditionalGeneration, WhisperProcessor
from peft import PeftModel

BASE_MODEL = "openai/whisper-small"
LORA_PATH = "./whisper-small-lora-ap"      
OUTPUT_MERGED = "./whisper-small-merged"   
device = "cpu"
model = WhisperForConditionalGeneration.from_pretrained(BASE_MODEL).to(device)
model = PeftModel.from_pretrained(model, LORA_PATH)
model = model.merge_and_unload()
model.save_pretrained(OUTPUT_MERGED)
processor = WhisperProcessor.from_pretrained(BASE_MODEL)
processor.save_pretrained(OUTPUT_MERGED)

print("Merged model saved to whisper-small-merged/")
```
